
# 🧠 Machine Learning Model Cheat Sheet

A compact, MSc-friendly reference for core ML models — their purpose, initialization, parameters, and tuning tips.  
Includes feature scaling guidance and evaluation metrics.

---



## ⚖️ Feature Scaling — When to Scale
| Model Type | Scaling Needed? | Recommended Scaler |
|-------------|------------------|--------------------|
| Linear / Logistic Regression | ✅ Yes | `StandardScaler` or `MinMaxScaler` |
| SVM (Regressor & Classifier) | ✅ Yes | `StandardScaler` |
| KNN | ✅ Yes | `StandardScaler` or `MinMaxScaler` |
| Decision Trees / Random Forest / XGBoost | ❌ No | Not required |
| KMeans / DBSCAN | ✅ Yes | `StandardScaler` or `MinMaxScaler` |

📘 *Scale when model relies on distance or gradient-based optimization.*



## 📊 Evaluation Metrics

| Task | Common Metrics | Notes |
|------|----------------|-------|
| **Regression** | MSE, RMSE, MAE, R² | Lower is better for errors; higher R² is better |
| **Binary Classification** | Accuracy, Precision, Recall, F1, ROC-AUC | Use F1/AUC for imbalance |
| **Multi-Class Classification** | Accuracy, Macro-F1 | Macro-F1 for imbalance |
| **Clustering** | Silhouette Score, Davies–Bouldin | Higher Silhouette = better |

## 📈 Regression Models

### Linear Regression
Predicts continuous outcomes assuming linear relationships.
```python
from sklearn.linear_model import LinearRegression
model = LinearRegression(fit_intercept=True, copy_X=True, n_jobs=None)
```
- `fit_intercept`: Include bias term (usually `True`)
- `n_jobs`: Use multiple cores
💡 Scale optional. Check residuals for linearity.

### Ridge / Lasso / ElasticNet
Regularized linear models (reduce overfitting).
```python
from sklearn.linear_model import Ridge, Lasso, ElasticNet
ridge = Ridge(alpha=1.0)
lasso = Lasso(alpha=0.1)
enet  = ElasticNet(alpha=1.0, l1_ratio=0.5)
```
- `alpha`: Regularization strength ↑ = more shrinkage
- `l1_ratio`: (ElasticNet only) mix of L1/L2 (0=L2, 1=L1)
💡 Scale before training. Tune with CV.

### SVM Regressor
Fits hyperplane to predict continuous values.
```python
from sklearn.svm import SVR
model = SVR(kernel='rbf', C=1.0, epsilon=0.1)
```
- `kernel`: `'linear'`, `'poly'`, `'rbf'`
- `C`: Regularization (↑ = less regularization)
- `epsilon`: Margin of tolerance  
💡 Scale data. Tune via grid search.

### Decision Tree Regressor
Splits data into regions to predict continuous outputs.
```python
from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor(max_depth=None, min_samples_split=2, random_state=42)
```
- `max_depth`: Limit to avoid overfitting
- `min_samples_split`: Min samples for a split
💡 No scaling needed. Prune or limit depth.

### Random Forest / XGBoost Regressor
Ensemble of trees for accuracy & robustness.
```python
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
rf  = RandomForestRegressor(n_estimators=100, max_depth=None, random_state=42)
xgb = XGBRegressor(n_estimators=100, learning_rate=0.1, max_depth=6)
```
- `n_estimators`: #trees (100–500 typical)
- `max_depth`: Control complexity
- `learning_rate`: (XGB) Step size shrinkage
💡 No scaling needed. Tune via search.



## 🧮 Classification Models

### Logistic Regression
Binary linear classifier, predicts probabilities.
```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(penalty='l2', C=1.0, solver='lbfgs', max_iter=1000, class_weight=None, random_state=42)
```
- `penalty`: Regularization (`'l2'`, `'l1'`, `'elasticnet'`)
- `C`: Inverse reg. strength (try 0.01–10)
- `solver`: Optimization algo (`'lbfgs'`, `'liblinear'` small data)
💡 Scale features. Great interpretable baseline.

### KNN Classifier
Classifies by majority vote of neighbors.
```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=5, weights='uniform', metric='minkowski')
```
- `n_neighbors`: 3–15 typical
- `weights`: `'uniform'` or `'distance'`
💡 Scale data. Larger k = smoother boundary.

### SVM Classifier
Finds max-margin hyperplane between classes.
```python
from sklearn.svm import SVC
model = SVC(kernel='rbf', C=1.0, gamma='scale', probability=True)
```
- `kernel`: `'linear'`, `'poly'`, `'rbf'`
- `C`: Regularization
- `gamma`: Curve tightness  
💡 Scale data. Tune `C` & `gamma`.

### Decision Tree Classifier
Splits data into pure nodes.
```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(max_depth=None, criterion='gini', random_state=42)
```
- `criterion`: `'gini'` or `'entropy'`
- `max_depth`: Control complexity  
💡 No scaling needed.

### Random Forest / XGBoost Classifiers
Ensemble methods for strong performance.
```python
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
rf  = RandomForestClassifier(n_estimators=100, max_depth=None, random_state=42)
xgb = XGBClassifier(n_estimators=100, learning_rate=0.1, max_depth=6)
```
💡 Same params as regressors. Use feature importance for insight.



## 🔍 Clustering Models

### K-Means
Partitions data into k clusters minimizing variance.
```python
from sklearn.cluster import KMeans
model = KMeans(n_clusters=5, init='k-means++', max_iter=300, random_state=42)
```
- `n_clusters`: Use Elbow/Silhouette method
- `init`: `'k-means++'` recommended  
💡 Scale data. Run multiple initializations.

### DBSCAN
Density-based, detects irregular clusters & noise.
```python
from sklearn.cluster import DBSCAN
model = DBSCAN(eps=0.5, min_samples=5, metric='euclidean')
```
- `eps`: Max neighbor distance
- `min_samples`: Points to form a dense cluster  
💡 Scale data. Great for noisy/non-spherical data.
